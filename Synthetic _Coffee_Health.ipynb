## Dataset Summary

[**Global Coffee Health Dataset**](https://www.kaggle.com/datasets/uom190346a/global-coffee-health-dataset):

The GlobalCoffeeHealth dataset contains 10,000 synthetic records reflecting real-world patterns of coffee consumption, sleep behavior, and health outcomes across 20 countries. It includes demographics, daily coffee intake, caffeine levels, sleep duration and quality, BMI, heart rate, stress, physical activity, health issues, occupation, smoking, and alcohol consumption.

The dataset captures realistic correlations observed in research—such as caffeine’s impact on sleep, stress, and health—making it ideal for statistical analysis, predictive modeling, and lifestyle or wellness studies.


*   **Scope**: Individual-level records linking coffee/caffeine habits, sleep, vitals, lifestyle, and self-reported health across multiple countries.
*   **Shape**: 10,000 rows × 16 columns
*   **Granularity**: One row per person (no time series).
*   **Keys**: ID is unique per row.
*   **Missingness**: None observed.
*   **Types**: Numeric (continuous + integer), categorical, and binary (0/1).

# Outliers

```python
numeric_features = df.select_dtypes(include=['int64', 'float64']).columns
categorical_features = df.select_dtypes(include='object').columns
```

```python
#Count of outliers

outlier_counts = {}
for col in numeric_features:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        count = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
        outlier_counts[col] = count
outlier_counts
```

```python
def remove_outliers_iqr(dataframe):
    df_clean = dataframe.copy()
    for col in numeric_features:
        Q1 = dataframe[col].quantile(0.25)
        Q3 = dataframe[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Keep only rows within bounds for this column
        df_clean = df_clean[(df_clean[col] >= lower_bound) & (df_clean[col] <= upper_bound)]

    return df_clean

df = remove_outliers_iqr(df)
df
```

```python
#Count of outliers after remove them

outlier_counts = {}
for col in numeric_features:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        count = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
        outlier_counts[col] = count
outlier_counts
```

```python
df
```

**For Outliers**:
- To check for outliers, I used the Interquartile Range (IQR) method.  
- The IQR is defined as the difference between the 75th percentile (Q3) and the 25th percentile (Q1).  
- Any value below Q1 - 1.5 * IQR or above Q3 + 1.5 * IQR is considered an outlier.  
- After detecting and count them, I remove them.

# Encoding

```python
categorical_features
```

```python
for col in categorical_features:
    print(df[col].value_counts(),"\n")
```

```python
df['Country'].unique()
```

```python
# Label Encoding for Gender, Country and Occupation

label_encoder = LabelEncoder()

for col in ['Gender', 'Country', 'Occupation']:
    df[col] = label_encoder.fit_transform(df[col])
```

```python
# Ordinal Encoding for Sleep_Quality, Stress_Level, Health_Issues

from sklearn.preprocessing import OrdinalEncoder

sleep_encoder = OrdinalEncoder(categories=[['Poor', 'Fair', 'Good', 'Excellent']])
stress_encoder = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
health_encoder = OrdinalEncoder(categories=[['No_Issue', 'Mild', 'Moderate', 'Severe']])

df['Sleep_Quality'] = sleep_encoder.fit_transform(df[['Sleep_Quality']])
df['Stress_Level'] = stress_encoder.fit_transform(df[['Stress_Level']])
df['Health_Issues'] = health_encoder.fit_transform(df[['Health_Issues']])
```

```python
df
```

- For **Gender, Country, and Occupation**: I used **Label Encoding** because these are nominal categorical features, meaning their values are just different categories without any natural order (e.g., Male, Female, Other or Egypt, USA, UK).  

- For **Sleep_Quality, Stress_Level, and Health_Issues**: I used **Ordinal Encoding** because these features have an inherent order.  
   - Sleep_Quality: Poor < Fair < Good < Excellent  
   - Stress_Level: Low < Medium < High  
   - Health_Issues: None < Mild < Moderate < Severe  
Ordinal encoding helps preserve this logical ranking when converting them into numbers.  


# Normalization

```python
scaler = StandardScaler()

df[numeric_features] = scaler.fit_transform(df[numeric_features])
```

- I applied **Normalization** to the numerical columns because their values have different ranges (for example: Age ranges from 18–80, Sleep_Hours from 3–10, and Caffeine_mg can go up to 1000).  

- Normalization rescales all numerical features into the same range, usually [0,1], so that no single feature dominates the analysis or the machine learning model.  

- This is especially useful when features are measured in different units (like hours, mg, and bpm), making the dataset more consistent and improving model performance.
